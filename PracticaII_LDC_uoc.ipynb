```python
import pandas as pd
import matplotlib as plt
```

```python
# cargar datos
file_path = "/content/Establiments_d_allotjament_tur_stic_inscrits_al_Registre_de_Turisme_de_Catalunya_20250609.csv"
df = pd.read_csv(file_path)
df
```

```python
# Exploración inicial - tipo de datos
print("Tipos de datos por columna")
print(df.dtypes)
```

```python
# resumen estadístico
print("\nResumen estadístico del DataFrame:")
print(df.describe(include='all'))
```

```python
# ver valores categóricos únicos
for col in df.select_dtypes(include='object').columns:
    print(f"Unique values in column '{col}':")
    print(df[col].unique())
    print()
```

```python
# Filtrar por tipo de modalidad
df = df[~df['Tipus establiment'].isin(['Càmping', 'Hotels'])]

# Borrar columnas no relevantes
# Lista de columnas a eliminar
columns_to_drop = [
    'Tipus de via', 'Nom de la via', 'Número', 'Pis', 'Porta',
    'Escala', 'Bloc', 'Dígit de control', 'Estat'
]

# Eliminar las columnas
df = df.drop(columns=columns_to_drop)
df
```

```python
# Eliminar duplicados
df = df.drop_duplicates()
df
```

```python
# missing values
print("Conteo de valores nulos por columna:")
print(df.isnull().sum())
```

```python
# código provincia
provincia_counts = df['Província'].value_counts()

# Guardar dataframe como csv
provincia_counts.to_csv('provincia_counts.csv', index=True)

provincia_counts
```

```python
# Guardar dataframe como csv
df.to_csv('dataset_establiments_clean.csv', index=False)
```

```python
# Filtrar filas donde 'Província' == 'Girona'
df_girona = df[df['Província'] == 'Girona']

# Contar número de alojamientos por municipio
municipi_counts = df_girona['Municipi'].value_counts().head(10).reset_index()

# Renombrar columnas
municipi_counts.columns = ['Municipi', 'Recompte']

# Guardar como CSV
municipi_counts.to_csv('municipi_counts_top10.csv', index=False)

# Mostrar el resultado
municipi_counts
```

```python
# Filtrar filas donde 'Província' == 'Girona'
df_tenidors = df[df['Província'] == 'Girona']

# Contar número de alojamientos por municipio
municipi_counts = df_girona['Municipi'].value_counts().head(10).reset_index()

# Renombrar columnas
municipi_counts.columns = ['Municipi', 'Recompte']

# Guardar como CSV
municipi_counts.to_csv('municipi_counts_top10.csv', index=False)

# Mostrar el resultado
municipi_counts
```

```python
# Agrupar por 'Província' y 'Nombre_completo' y contar
df['Nombre_completo'] = df['Nom del titular']+ '-' + df['Primer cognom'] + '-' + df['Segon cognom']
df = df[df['Nombre_completo'] != 'No aplica-No aplica-No aplica']

grouped = df.groupby(['Província', 'Nombre_completo']).size().reset_index(name='Counts')

# Calcular el total por provincia
totals = grouped.groupby('Província')['Counts'].transform('sum')

# Calcular el porcentaje por propietario respecto al total de su provincia
grouped['Porcentaje'] = (grouped['Counts'] / totals * 100).round(2)

# Reordenar columnas
result = grouped[['Nombre_completo', 'Counts', 'Província', 'Porcentaje']]

# Guardar como CSV
result.to_csv('propietarios_por_provincia.csv', index=False)

# Mostrar resultado
result
```

```python
# Agrupar por 'Província' y 'Nombre_completo' y contar
grouped = df.groupby(['Província', 'Nombre_completo']).size().reset_index(name='Counts')

# Calcular el total por provincia para el porcentaje
grouped['Total_provincia'] = grouped.groupby('Província')['Counts'].transform('sum')
grouped['Porcentaje'] = (grouped['Counts'] / grouped['Total_provincia'] * 100).round(2)

# Eliminar columna auxiliar
grouped = grouped.drop(columns='Total_provincia')

# Reordenar columnas
result = grouped[['Nombre_completo', 'Counts', 'Província', 'Porcentaje']]

# Obtener lista única de provincias
provincias = result['Província'].unique()

# Guardar un CSV para cada provincia con solo el top 10
for prov in provincias:
    df_prov = result[result['Província'] == prov].sort_values(by='Counts', ascending=False).head(10)

    # Mostrar en consola
    print(f"🏙️ Top 10 propietarios en {prov}")
    print(df_prov, "\n")

    # Guardar CSV
    filename = f'propietarios_top10_{prov}.csv'.replace(" ", "_")
    df_prov.to_csv(filename, index=False)
```

```python
# Agrupar por 'Província' y 'Nombre_completo' y contar
grouped = df.groupby(['Província', 'Nombre_completo']).size().reset_index(name='Counts')

# Calcular el total por provincia para el porcentaje
grouped['Total_provincia'] = grouped.groupby('Província')['Counts'].transform('sum')
grouped['Porcentaje'] = (grouped['Counts'] / grouped['Total_provincia'] * 100).round(2)

# Eliminar columna auxiliar
grouped = grouped.drop(columns='Total_provincia')

# Reordenar columnas
result = grouped[['Nombre_completo', 'Counts', 'Província', 'Porcentaje']]

# Obtener lista única de provincias
provincias = result['Província'].unique()

# Lista para guardar DataFrames de cada top10 provincial
top10_list = []

for prov in provincias:
    df_prov = result[result['Província'] == prov].sort_values(by='Counts', ascending=False).head(10)
    top10_list.append(df_prov)

# Concatenar todos los top10 en un solo DataFrame
df_top10_all = pd.concat(top10_list)

# Guardar todo en un solo CSV
df_top10_all.to_csv('propietarios_top10_todas_provincias.csv', index=False)

df_top10_all
```
